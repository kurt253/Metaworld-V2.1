# Export registers naar Excel

Leest de opgeslagen JSON-bestanden en schrijft een Excel-bestand met:
- **Tabblad Rijksregister**: rijksregisternummer, familienaam, voornaam
- **Tabblad Bisregister**: bisnummer, familienaam, voornaam

```python
import pandas as pd
from pathlib import Path

ROOT     = Path().resolve().parent if Path().resolve().name == 'Notebooks' else Path().resolve()
SAVE_DIR = ROOT / 'Generated Data'

# ── Pas PREFIX aan om een andere dataset te gebruiken ────────────────────────
PREFIX = 'metaworld'

print(f'ROOT     : {ROOT}')
print(f'SAVE_DIR : {SAVE_DIR}')
print(f'PREFIX   : {PREFIX}')
```

```python
rr_pad  = SAVE_DIR / f'{PREFIX}_rijksregister.json'
bis_pad = SAVE_DIR / f'{PREFIX}_bisregister.json'

for pad in [rr_pad, bis_pad]:
    if not pad.exists():
        raise FileNotFoundError(f'Bestand niet gevonden: {pad}\nGenereer eerst de data via voorbereiding_data.ipynb.')

rr  = pd.read_json(rr_pad,  convert_dates=False)
bis = pd.read_json(bis_pad, convert_dates=False)

print(f'Rijksregister : {len(rr):,} personen')
print(f'Bisregister   : {len(bis):,} personen')
```

```python
rr_export  = rr[['rijksregisternummer', 'familienaam', 'voornaam']].copy()
bis_export = bis[['bisnummer', 'familienaam', 'voornaam']].copy()

print('Rijksregister (eerste 3 rijen):')
display(rr_export.head(3))

print('Bisregister (eerste 3 rijen):')
display(bis_export.head(3))
```

```python
xls_pad = SAVE_DIR / f'{PREFIX}_namen.xlsx'

with pd.ExcelWriter(xls_pad, engine='openpyxl') as writer:
    rr_export.to_excel(writer,  sheet_name='Rijksregister', index=False)
    bis_export.to_excel(writer, sheet_name='Bisregister',   index=False)

print(f'Opgeslagen: {xls_pad}')
print(f'  Tabblad Rijksregister : {len(rr_export):,} rijen')
print(f'  Tabblad Bisregister   : {len(bis_export):,} rijen')
```
